## 🔧 Converting TotalCharges into Numerical Format

The **TotalCharges** column contains the total amount paid by customers.

Since this column represents numerical values, it should be converted from object type to float type for accurate calculations and analysis.

Before conversion, we will check for any blank values present in this column.

```python
# Checking blank values in TotalCharges column

df[df['TotalCharges'] == ' '].shape
```

### Observation

The **TotalCharges** column contains **11 blank values**.

Since these values cannot be directly converted into numerical format, they need to be handled before changing the data type.

We will replace these blank values with `NaN` and then convert the column into a numerical format.

```python
# Replacing blank values with NaN

df['TotalCharges'] = df['TotalCharges'].replace(' ', np.nan)

# Checking missing values after replacement

df['TotalCharges'].isnull().sum()
```

### Observation

The blank values in the **TotalCharges** column have been successfully converted into missing values (`NaN`).

Now, we can safely convert this column from object type to numerical format.

```python
# Converting TotalCharges column into numerical format

df['TotalCharges'] = df['TotalCharges'].astype(float)

# Checking updated data type

df['TotalCharges'].dtype
```

### Observation

The **TotalCharges** column has been successfully converted from **object data type to float64**.

This conversion allows us to perform numerical calculations and statistical analysis on total customer charges.

The remaining missing values will be handled before starting Exploratory Data Analysis.

## 🔍 Handling Missing Values

## 🎯 Understanding the Target Variable

The target variable of this dataset is **Churn**.

- **Yes** → Customer has left the service
- **No** → Customer is still using the service

First, we will analyze the distribution of churned and retained customers.

```python
# Checking churn value counts

df['Churn'].value_counts()
```

```python
# Checking churn percentage

df['Churn'].value_counts(normalize=True)*100
```

```python
# Visualizing customer churn distribution

plt.figure(figsize=(7,5))

ax = sns.countplot(
    data=df,
    x='Churn'
)

plt.title("Customer Churn Distribution", fontsize=14, fontweight='bold')
plt.xlabel("Churn Status")
plt.ylabel("Number of Customers")

for container in ax.containers:
    ax.bar_label(container)

plt.show()
```

## 📌 Observation

The customer churn distribution shows that:

- **5163 customers (73.42%) are retained** and continue using the service.
- **1869 customers (26.58%) have churned** and discontinued the service.

The graph indicates that although most customers are retained, more than one-fourth of customers have left the company.

## 💡 Business Insight

A churn rate of **26.58%** is significant for the business.

The company should analyze customer behavior, service usage, contract types, and payment patterns to identify the major reasons behind customer churn.

Understanding these factors can help the company design better customer retention strategies.

# 🔥 Correlation Analysis

Correlation analysis helps identify relationships between different features in the dataset.

Since correlation can only be calculated on numerical values, categorical variables will first be converted into numerical form using Label Encoding.

The correlation matrix will help identify which features have the strongest relationship with customer churn.

```python
# Creating a copy of dataset for correlation analysis

df_corr = df.copy()
```

```python
# Importing LabelEncoder

from sklearn.preprocessing import LabelEncoder
```

```python
# Encoding all categorical columns

le = LabelEncoder()

for column in df_corr.columns:
    if df_corr[column].dtype == 'object':
        df_corr[column] = le.fit_transform(df_corr[column])
```

```python
# Checking encoded dataset

df_corr.head()
```

## 📌 Observation

The dataset has been successfully encoded using **Label Encoding**, where all categorical variables have been converted into numerical values.

This transformation is required because correlation analysis can only be performed on numerical data.

The encoded dataset retains the same information as the original dataset while making it suitable for calculating correlations between features.

```python
# Calculating correlation matrix

corr_matrix = df_corr.corr()

corr_matrix
```
